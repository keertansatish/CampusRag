```python
from pathlib import Path

from langchain_community.document_loaders import PyMuPDFLoader

pdf_directory = Path.cwd() / "RAG"
if not pdf_directory.is_dir():
    raise FileNotFoundError(f"PDF directory does not exist: {pdf_directory}")

pdf_files = sorted(
    path for path in pdf_directory.rglob("*")
    if path.is_file() and path.suffix.lower() == ".pdf"
)
if not pdf_files:
    raise FileNotFoundError(f"No PDF files found in {pdf_directory}")

documents = []
for pdf_file in pdf_files:
    documents.extend(PyMuPDFLoader(pdf_file, mode="page").load())

print(f"Loaded {len(documents)} PDF pages from {pdf_directory}")
```
